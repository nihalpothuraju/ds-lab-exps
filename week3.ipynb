```python

import numpy as np
import pandas as pd
from scipy.spatial import distance
from scipy.stats import spearmanr
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

A = np.array([2, 4, 6])
B = np.array([5, 1, 9])

d = distance.euclidean(A, B)
print("Euclidean Distance:", d)
print("Euclidean Similarity:", 1 / (1 + d))

d = distance.cityblock(A, B)
print("\nManhattan Distance:", d)
print("Manhattan Similarity:", 1 / (1 + d))

d = distance.minkowski(A, B, p=3)
print("\nMinkowski Distance (p=3):", d)
print("Minkowski Similarity:", 1 / (1 + d))

df = pd.DataFrame({
    'X': [10, 20, 30, 40, 50],
    'Y': [12, 24, 33, 45, 60]
})
print("\nPearson Correlation Matrix:")
print(df.corr(method='pearson'))

df = pd.DataFrame({
    'X': [10, 20, 30, 40, 50],
    'Y': [12, 18, 33, 47, 55]
})
corr, p = spearmanr(df['X'], df['Y'])
print("\nSpearman Correlation:", corr)
print("P-value:", p)

def hamming_distance(s1, s2):
    if len(s1) != len(s2):
        raise ValueError("Strings must be of equal length")
    return sum(a != b for a, b in zip(s1, s2))

print("\nHamming Distance:",
      hamming_distance("karolin", "kathrin"))

def jaccard_index(s1, s2):
    set1, set2 = set(s1.split()), set(s2.split())
    union = set1 | set2
    if not union:
        return 1.0
    return len(set1 & set2) / len(union)

s1 = "data science is fun"
s2 = "science makes data useful"
print("\nJaccard Index:", jaccard_index(s1, s2))

vectorizer = CountVectorizer().fit([s1, s2])
vectors = vectorizer.transform([s1, s2])
cos = cosine_similarity(vectors[0], vectors[1])[0][0]
print("Cosine Similarity:", cos)

def lcs_length(X, Y):
    m, n = len(X), len(Y)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m):
        for j in range(n):
            if X[i] == Y[j]:
                dp[i + 1][j + 1] = dp[i][j] + 1
            else:
                dp[i + 1][j + 1] = max(
                    dp[i][j + 1], dp[i + 1][j]
                )
    return dp[m][n]

print("\nLCS Length:",
      lcs_length("ABCDEF", "AEBDF"))
```
