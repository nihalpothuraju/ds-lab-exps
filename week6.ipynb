```python

import pandas as pd
from sklearn.datasets import load_iris
from scipy.stats import spearmanr, chi2_contingency

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print("Pearson Correlation:")
print(df.corr(method="pearson").round(3))

corr, p = spearmanr(
    df["sepal length (cm)"],
    df["petal length (cm)"]
)
print("\nSpearman Correlation:", round(corr, 3))
print("P-value:", p)

table = pd.crosstab(iris.target, df["sepal width (cm)"] > 3)
chi2, p, dof, expected = chi2_contingency(table)

print("\nChi-square Statistic:", round(chi2, 3))
print("P-value:", round(p, 4))
```

```python

import numpy as np
from scipy.stats import ttest_ind, f_oneway, norm
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target

a = X[y == 0, 0]
b = X[y == 1, 0]

t_stat, p_t = ttest_ind(a, b)
print("T-test Statistic:", round(t_stat, 3))
print("T-test P-value:", p_t)

groups = [X[y == i, 0] for i in range(3)]
f_stat, p_a = f_oneway(*groups)

print("\nANOVA F-statistic:", round(f_stat, 3))
print("ANOVA P-value:", p_a)

sample = np.array([52, 54, 51, 53, 55])
population_mean = 50
population_std = 4

z = (sample.mean() - population_mean) / (
    population_std / np.sqrt(len(sample))
)
p_z = 2 * norm.sf(abs(z))

print("\nZ-test Statistic:", round(z, 3))
print("Z-test P-value:", round(p_z, 4))
```

```python

import pandas as pd
from scipy.stats import zscore

df = pd.DataFrame({
    "Value": [10, 12, 11, 13, 12, 14, 100]
})

df["Z-score"] = zscore(df["Value"])
print("Z-score Outliers:")
print(df[df["Z-score"].abs() > 3])

Q1 = df["Value"].quantile(0.25)
Q3 = df["Value"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("\nIQR Outliers:")
print(df[(df["Value"] < lower) |
         (df["Value"] > upper)])
```

```python

import numpy as np
from scipy.stats import t

data = np.array([10, 12, 11, 13, 12, 14, 100])

n = len(data)
mean = np.mean(data)
std = np.std(data, ddof=1)

G = np.max(np.abs(data - mean)) / std

alpha = 0.05
t_critical = t.ppf(1 - alpha / (2 * n), n - 2)
G_critical = ((n - 1) / np.sqrt(n)) * np.sqrt(
    t_critical**2 / (n - 2 + t_critical**2)
)

print("Grubbs Statistic:", round(G, 3))
print("Critical Value:", round(G_critical, 3))

if G > G_critical:
    print("Outlier detected")
else:
    print("No outlier detected")
```
